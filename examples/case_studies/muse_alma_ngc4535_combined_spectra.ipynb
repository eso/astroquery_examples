**The story:** A Galaxy catches our eye in [Rubin First Look](https://rubinobservatory.org/gallery/image-releases/rubin-first-look/cosmic-treasure-chest). What else can we learn about it using Astroquery and Virtual Observatory services? ## Setup · ESO and ALMA SODA cutout helpers

````python
def _soda_request(service, dp_id, ra=None, dec=None, radius=None,
                  wave_min=None, wave_max=None, prefix=None):
    spatial = (ra, dec, radius)
    if any(x is None for x in spatial) and not all(x is None for x in spatial):
        raise ValueError("ra, dec, and radius must be supplied together")
    if (wave_min is None) != (wave_max is None):
        raise ValueError("wave_min and wave_max must be supplied together")
    if radius is not None and float(radius) <= 0:
        raise ValueError("radius must be positive")
    if wave_min is not None and float(wave_min) > float(wave_max):
        raise ValueError("wave_min must be <= wave_max")
    params = []
    if service == "muse":
        base = "https://dataportal.eso.org/dataPortal/soda/sync"
        params.append(("ID", str(dp_id)))
        if radius is not None:
            params.append(("CIRCLE", f"{float(ra)} {float(dec)} {float(radius):.6f}"))
    elif service == "alma":
        base = "https://almascience.eso.org/soda/sync"
        params.extend((("REQUEST", "queryData"), ("ID", str(dp_id))))
        if radius is not None:
            params.append(("POS", f"CIRCLE {float(ra)} {float(dec)} {float(radius):.6f}"))
    else:
        raise ValueError(f"Unknown cutout service: {service}")
    if wave_min is not None:
        params.append(("BAND", f"{float(wave_min)} {float(wave_max)}"))
    if prefix is not None:
        params.append(("PREFIX", str(prefix)))
    return f"{base}?{urlencode(params)}"


def _content_length(url, timeout=600):
    try:
        with requests.head(url, timeout=timeout, allow_redirects=True) as response:
            if response.ok and response.headers.get("Content-Length"):
                return int(response.headers["Content-Length"])
        with requests.get(url, timeout=timeout, stream=True) as response:
            response.raise_for_status()
            length = response.headers.get("Content-Length")
            return int(length) if length else None
    except requests.RequestException:
        return None


def _size_label(n):
    if n is None: return "unknown"
    size = float(n)
    for unit in ("B", "KB", "MB", "GB", "TB"):
        if size < 1000 or unit == "TB": return f"{size:.1f} {unit}"
        size /= 1000


def _download_soda(url, path, *, timeout=600, overwrite=False, show_progress=True,
                   print_size=True, progress_backend="tqdm"):
    if progress_backend not in ("tqdm", "astropy", "none"):
        raise ValueError("progress_backend must be 'tqdm', 'astropy', or 'none'")
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    if path.exists() and not overwrite:
        print(f"{path} exists; skipping download")
        return str(path)
    if print_size:
        print("Estimated download size:", _size_label(_content_length(url, timeout)))
    if progress_backend == "astropy":
        temporary = Path(download_file(url, cache=False,
            show_progress=show_progress, timeout=timeout))
        shutil.copyfile(temporary, path)
        return str(path)
    temporary_name = None
    try:
        with requests.get(url, stream=True, timeout=timeout) as response:
            response.raise_for_status()
            total = int(response.headers.get("Content-Length", 0)) or None
            with tempfile.NamedTemporaryFile(dir=path.parent, delete=False) as tmp:
                temporary_name = tmp.name
            progress = (tqdm(total=total, unit="B", unit_scale=True,
                             desc="Downloading")
                        if progress_backend == "tqdm" and tqdm is not None and show_progress else None)
            try:
                with open(temporary_name, "wb") as stream:
                    for chunk in response.iter_content(chunk_size=1024*1024):
                        if chunk:
                            stream.write(chunk)
                            if progress is not None: progress.update(len(chunk))
            finally:
                if progress is not None: progress.close()
        os.replace(temporary_name, path)
        temporary_name = None
    finally:
        if temporary_name is not None:
            Path(temporary_name).unlink(missing_ok=True)
    return str(path)


def _get_cutout(service, dp_id, ra=None, dec=None, radius=None,
                wave_min=None, wave_max=None, prefix=None, *,
                outdir=None, outfile=None, verbose=False, download=True,
                show_progress=True, timeout=600, overwrite=False,
                print_size=True, progress_backend="tqdm"):
    url = _soda_request(service, dp_id, ra, dec, radius,
                        wave_min, wave_max, prefix)
    if verbose: print("SODA request:", url)
    if not download:
        if print_size:
            print("Server reports size:", _size_label(_content_length(url, timeout)))
        return url
    destination = Path(outdir) if outdir is not None else Path.cwd()
    filename = Path(str(outfile or f"{dp_id}.fits")).name
    filename = "".join("_" if c in r'\\/:*?"<>|' else c for c in filename)
    path = _download_soda(url, destination/filename, timeout=timeout,
                          overwrite=overwrite, show_progress=show_progress,
                          print_size=print_size, progress_backend=progress_backend)
    if verbose: print("Wrote:", path)
    return path


def muse_cutout(dp_id, ra=None, dec=None, radius=None,
                 wave_min=None, wave_max=None, prefix=None, **kwargs):
    """ESO SODA: ID, optional CIRCLE (degrees), optional BAND (metres)."""
    return _get_cutout("muse", dp_id, ra, dec, radius,
                       wave_min, wave_max, prefix, **kwargs)


def alma_cutout(dp_id, ra=None, dec=None, radius=None,
                 wave_min=None, wave_max=None, prefix=None, **kwargs):
    """ALMA SODA: REQUEST=queryData, ID, optional POS and BAND (metres)."""
    return _get_cutout("alma", dp_id, ra, dec, radius,
                       wave_min, wave_max, prefix, **kwargs)


def get_wavelengthaxis(hdul, index=1):
    """Linear FITS spectral axis in the header's CUNIT3 (MUSE Å, ALMA Hz here)."""
    h = hdul[index].header
    delta = h.get("CD3_3", h.get("CDELT3"))
    if delta is None:
        raise ValueError("No linear third-axis increment in FITS header")
    return h["CRVAL3"] + (np.arange(h["NAXIS3"])+1-h["CRPIX3"])*delta
````

## Setup · reusable displays and data wrangling

````python
def display_linked_sky(viewers, labels, columns=2):
    """Link existing Aladin widgets and arrange their labelled panels."""
    if len(viewers) != len(labels):
        raise ValueError("One label is required per Aladin viewer")
    for viewer in viewers[1:]:
        widgets.jslink((viewers[0], "_target"), (viewer, "_target"))
        widgets.jslink((viewers[0], "_fov"), (viewer, "_fov"))
    panels = [VBox([HTML(f"<b>{label}</b>"), viewer])
              for label, viewer in zip(labels, viewers)]
    display(GridBox(panels, layout=Layout(display="grid",
        grid_template_columns=f"repeat({columns}, minmax(0, 1fr))", grid_gap="10px")))


def display_galaxy_properties(s, n):
    if s is None or n is None or not len(s) or not len(n):
        raise RuntimeError("SIMBAD or NED did not return the galaxy.")
    print("SIMBAD")
    display(s[[c for c in ("main_id", "ra", "dec", "otype", "morph_type",
                              "rvz_redshift", "rvz_radvel", "galdim_majaxis")
               if c in s.colnames]])
    print("NED")
    display(n[[c for c in ("Object Name", "Type", "Velocity", "Redshift",
                              "Magnitude and Filter") if c in n.colnames]])
    return s["rvz_radvel"].quantity[0].to(u.km/u.s)

def display_publications(papers, phangs_papers, limit=10):
    """Show compact ADS result tables and overlay both publication histories."""
    for label, results in (("All papers", papers), ("PHANGS papers", phangs_papers)):
        if results is None:
            raise ValueError(f"No ADS result for {label}")
        rows = []
        for paper in reversed(results[-limit:]):
            title = paper["title"][0] if len(paper["title"]) else "Untitled"
            authors = ", ".join(str(x) for x in paper["author"][:3])
            if len(paper["author"]) > 3:
                authors += ", et al."
            rows.append((str(paper["pubdate"])[:4], title, authors,
                         str(paper["bibcode"]), paper["citation_count"]))
        print(f"{label}: {len(results)} results (showing {len(rows)})")
        display(Table(rows=rows, names=("Year", "Title", "Authors", "Bibcode", "Citations")))
    if len(papers):
        years = np.array([int(str(y)[:4]) for y in papers["pubdate"]])
        phangs_years = np.array([int(str(y)[:4]) for y in phangs_papers["pubdate"]])
        bins = np.arange(years.min()-.5, years.max()+1.5, 1)
        fig, ax = plt.subplots(figsize=(10, 4))
        ax.hist(years, bins=bins, color="C0", edgecolor="black", label="All papers")
        ax.hist(phangs_years, bins=bins, color="C1", edgecolor="black",
                alpha=.85, label="PHANGS papers")
        ax.set(xlabel="Publication year", ylabel="Number of papers",
               title="ADS papers associated with NGC 4535")
        ax.legend(); fig.tight_layout(); plt.show()


def display_eso_products(products):
    if products is None or not len(products):
        raise RuntimeError("No ESO PHANGS products found")
    print("ESO PHANGS products:", len(products))
    display(products[[c for c in ("dp_id", "instrument_name", "dataproduct_subtype", "access_url")
                      if c in products.colnames]][:25])

def observed_window(rest, velocity, half_width):
    a = (velocity-half_width).to(u.m, equivalencies=u.doppler_optical(rest)).value
    b = (velocity+half_width).to(u.m, equivalencies=u.doppler_optical(rest)).value
    return min(a,b), max(a,b)


def spatial_mean(cube):
    """Mean over sky pixels, retaining NaN for empty spectral channels."""
    valid = np.isfinite(cube)
    count = valid.sum(axis=(-2, -1))
    if not np.any(count):
        raise ValueError("Cutout contains no finite pixels; check the position and FITS cube.")
    total = np.where(valid, cube, 0).sum(axis=(-2, -1))
    return np.divide(total, count, out=np.full(count.shape, np.nan), where=count > 0)


def spectral_median(cube):
    """Median at populated sky pixels; leave uncovered pixels masked."""
    if cube.ndim != 3:
        raise ValueError("Expected a spectral cube with shape (wavelength, y, x).")
    usable = np.isfinite(cube).any(axis=0)
    if not np.any(usable):
        raise ValueError("MUSE cutout contains no finite pixels; check its spectral band.")
    continuum = np.full(cube.shape[1:], np.nan)
    continuum[usable] = np.nanmedian(cube[:, usable], axis=0)
    return continuum


def display_halpha_maps(path):
    with fits.open(path) as h:
        cube = np.asarray(h[1].data, dtype=float)
        continuum = spectral_median(cube)
        valid = np.isfinite(cube) & np.isfinite(continuum)[None, :, :]
        line = np.where(valid, cube - continuum[None, :, :], 0).sum(axis=0)
        line[~valid.any(axis=0)] = np.nan
    coverage = np.isfinite(cube).any(axis=0)
    print(f"MUSE cube: {cube.shape}; finite voxels: {np.isfinite(cube).mean():.1%}; "
          f"covered sky pixels: {coverage.mean():.1%}")
    fig, axes = plt.subplots(1, 2, figsize=(10, 6))
    for ax, image, title, palette in zip(
        axes, (line, continuum), ("H-alpha Emission Map", "Continuum Map"),
        ("inferno", "viridis")):
        shown = np.sqrt(np.abs(image))
        cmap = plt.get_cmap(palette).copy(); cmap.set_bad(cmap(0))
        ax.imshow(shown, origin="lower", cmap=cmap,
                  vmin=0, vmax=np.nanpercentile(shown, 99.5))
        ax.set(title=title, xlabel="X (pixels)", ylabel="Y (pixels)")
    fig.tight_layout(); plt.show()
    return line, continuum

def display_nebula_catalog(table):
    if table is None or not len(table):
        raise RuntimeError("No PHANGS nebula catalogue rows")
    print("H II regions:", len(table))
    display(table[[c for c in ("GAL_NAME", "RAJ2000", "DECJ2000", "HA6562_FLUX")
                   if c in table.colnames]][:10])

def auxiliary_mask_id(links):
    table = links.to_table()
    auxiliary = table[table["semantics"] == "#auxiliary"]
    if not len(auxiliary):
        print("No auxiliary nebula mask advertised")
        return None
    return str(auxiliary["access_url"][0]).replace(
        "https://dataportal.eso.org/dataPortal/file/", "")

def display_nebula_mask(local, line=None, continuum=None):
    path = local[0] if isinstance(local, (list, tuple)) else local
    with fits.open(path) as h:
        mask = np.asarray(h[0].data, dtype=float)
    layers = ((line, "H-alpha Emission Map", "inferno"),
              (continuum, "Continuum Map", "viridis"),
              (mask, "Mask Map", "turbo")) if line is not None else ((mask,"Mask Map","turbo"),)
    fig, axes = plt.subplots(1, len(layers), figsize=(10, 6))
    if len(layers) == 1: axes = [axes]
    for ax, (layer, title, palette) in zip(axes, layers):
        shown = np.sqrt(np.abs(layer))
        cmap = plt.get_cmap(palette).copy(); cmap.set_bad(cmap(0))
        ax.imshow(shown, origin="lower", cmap=cmap,
                  vmin=0, vmax=np.nanpercentile(shown, 99.5))
        ax.set(title=title, xlabel="X (pixels)", ylabel="Y (pixels)")
    fig.tight_layout(); plt.show()
    return Path(path)

def ranked_nebula_positions(catalog):
    order = np.argsort(np.ma.filled(catalog["HA6562_FLUX"], -np.inf))[::-1]
    for index in order:
        row = catalog[index]
        yield index, SkyCoord(float(row["RAJ2000"])*u.deg,
                              float(row["DECJ2000"])*u.deg)

def brightest_nebula_position(catalog):
    index = np.argmax(catalog["HA6562_FLUX"])
    row = catalog[index]
    position = SkyCoord(row["RAJ2000"] * u.deg, row["DECJ2000"] * u.deg)
    return index, position

def co_covering_rows(rows, expected):
    if rows is None or not len(rows):
        return None
    centre = rows["frequency"].quantity.to(u.GHz)
    half = rows["bandwidth"].quantity.to(u.GHz)/2
    matches = rows[(centre-half<=expected) & (expected<=centre+half)]
    return matches if len(matches) else None

def display_muse_region_spectrum(path, systemic):
    with fits.open(path) as h:
        wave = get_wavelengthaxis(h) * u.Angstrom
        spectrum = spatial_mean(np.asarray(h[1].data,dtype=float))
    print(f"MUSE regional spectrum: {np.isfinite(spectrum).sum()}/{len(spectrum)} finite channels")
    fig, ax = plt.subplots(figsize=(10,6))
    ax.plot(wave.value, spectrum, label="Full spectrum for region of interest")
    ax.fill_between(wave.value, 0, spectrum,
                    where=np.isfinite(spectrum) & (spectrum > 0), alpha=0.3, color="C0")
    for label, rest in (("Hβ",4861),("[O III]",5007),("Hα",6563),("[S II]",6731)):
        observed = systemic.to(u.Angstrom,
                    equivalencies=u.doppler_optical(rest*u.Angstrom)).value
        ax.axvline(observed, color="0.6", ls="--", alpha=.6)
        ax.text(observed+15, np.nanmax(spectrum)*.7, label, rotation=90, color="0.4")
    ax.set(xlim=(4600,9100), xlabel="Wavelength (Å)", ylabel="Mean flux density",
           title="Full MUSE spectrum for selected region")
    ax.legend(); fig.tight_layout(); plt.show()
    return spectrum

def display_co_transition(transitions, velocity):
    if transitions is None or not len(transitions):
        raise RuntimeError("Splatalogue returned no CO(2-1) transition")
    rest = float(transitions["orderedfreq"][0]) * u.MHz
    observed = velocity.to(u.GHz, equivalencies=u.doppler_optical(rest))
    print(f"CO(2–1): rest {rest.to(u.GHz):.6f}; observed {observed:.6f}")
    return rest, observed

def select_alma_science_cube(rows, info):
    if info is None or not len(info):
        raise RuntimeError("ALMA DataLink returned no file metadata")
    urls = [str(x) for x in info["access_url"]]
    candidates = [x for x in urls if x.endswith(".cube.I.pbcor.fits") and ".ari_l." in x]
    pattern = re.compile(r"_(\d+)MHz\.12m\.cube\.I\.pbcor\.fits$")
    choices = []
    for i, row in enumerate(rows):
        token = str(row["member_ous_uid"]).replace("uid://","uid___").replace("/","_")
        target = str(row["target_name"])
        frequency = rows["frequency"].quantity[i].to_value(u.GHz)
        for url in candidates:
            match = pattern.search(url)
            if match and f"member.{token}.ari_l.{target}_sci." in url:
                offset = abs((int(match.group(1))*u.MHz).to_value(u.GHz)-frequency)
                choices.append((offset, url, str(row["proposal_id"])))
    if not choices: raise RuntimeError("No matching ALMA ARI-L primary-beam-corrected science cube")
    _, url, proposal = min(choices, key=lambda item:item[0])
    print("ALMA proposal:", proposal, "product:", url)
    return url

def alma_dataset_id(url):
    return Path(urlparse(url).path).name

def gas_comparison(muse_path, alma_path, systemic, ha_rest=HA_REST, co_rest=CO21_REST):
    with fits.open(muse_path) as m, fits.open(alma_path) as a:
        print("MUSE units:",m[1].header.get("BUNIT"),"ALMA units:",a[0].header.get("BUNIT"))
        mwave = get_wavelengthaxis(m)*u.Angstrom
        mf = spatial_mean(np.asarray(m[1].data,dtype=float))
        afreq = get_wavelengthaxis(a,index=0)*u.Hz
        acube = np.asarray(a[0].data,dtype=float)
        if acube.ndim==4: acube=acube[0]
        if acube.ndim!=3: raise ValueError("Unexpected ALMA cube axes; inspect FITS header")
        af = spatial_mean(acube)
    mv = mwave.to(u.km/u.s,equivalencies=u.doppler_optical(ha_rest))-systemic
    av = afreq.to(u.km/u.s,equivalencies=u.doppler_optical(co_rest))-systemic
    print(f"Spectral coverage: MUSE {np.isfinite(mf).sum()}/{len(mf)}, "
          f"ALMA {np.isfinite(af).sum()}/{len(af)} finite channels")
    def normalize(x):
        peak = np.nanmax(x)
        if not np.isfinite(peak) or peak<=0: raise ValueError("No positive line peak")
        return x/peak
    mnorm, anorm = normalize(mf), normalize(af)
    fig,ax=plt.subplots(figsize=(10,6))
    ax.plot(mv.value,mnorm,label="MUSE Hα",color="C0")
    ax.fill_between(mv.value,-1,mnorm,alpha=.2,color="C0")
    ax.plot(av.value,anorm,label="ALMA CO(2-1)",color="C1")
    ax.fill_between(av.value,-1,anorm,alpha=.2,color="C1")
    ax.axvline(0,color="0.6",ls="--")
    ax.set(xlim=(-150,150),ylim=(-.1,1.1),xlabel="Velocity relative to systemic (km/s)",
           ylabel="Independently peak-normalized signal",
           title="Normalised spectra for NGC4535")
    ax.legend();fig.tight_layout();plt.show()
    return fig
````

## 2 · Which galaxy is it, and what do we already know?

A name resolves to a sky coordinate. SIMBAD and NED supply identifiers, morphology and recession velocity. The velocity sets the observed positions of Hα and CO.

````python
"""Query SIMBAD and NED with Astroquery for the target's properties."""
coords = SkyCoord.from_name(TARGET)
sim = Simbad()
sim.add_votable_fields("oid", "otype", "morphtype", "rvz_redshift", "rvz_radvel", "dim")
s = sim.query_object(TARGET)
n = Ned.query_object(TARGET)

velocity = display_galaxy_properties(s, n)
print("Shared sky position:", coords.to_string("hmsdms"))
print("Systemic velocity:", velocity)
````

## 3 · What has already been published?

NASA ADS finds papers associated with the galaxy, then narrows to work mentioning PHANGS. This is background for the data search, rather than the central result.

````python
"""Build the resolved-object ADS query explicitly and search it."""
ads_simbad_id = str(s["oid"][0])
ads_ned_id = str(n["Object Name"][0]).replace(" ", "_")

ads_query = (
    f'((=abs:"{TARGET}" OR simbid:"{ads_simbad_id}" OR '
    f'nedid:"{ads_ned_id}") database:astronomy)'
)

print("ADS query:", ads_query)
ADS.NROWS = 1000
papers = ADS.query_simple(ads_query)
print("All ADS papers:", len(papers))
````

````python
"""Restrict ADS to PHANGS, then overlay both publication histories."""
phangs_query = f'({ads_query} AND full:"PHANGS")'
print("ADS PHANGS query:", phangs_query)
phangs_papers = ADS.query_simple(phangs_query)

display_publications(papers, phangs_papers)
````

## 4 · What does ESO have?

## Takeaway

Observations of the same region can be found in both **ESO** and **ALMA** archives via services that implement **VO standards**. Their cube products can be used to study different physical processes in the same region. The combined spectra turn a beautiful galaxy image into a qualitative and semi-quantitative description of ionized and molecular gas.

 - **Aladin** displays a region across several surveys, providing links to astronomical data;
 - **SIMBAD** identifies objects and supplies catalogue properties;
 - **Astroquery** queries ESO and ALMA astronomical data via **TAP/ADQL** under the hood;
 - **VO DataLink** reveals associated products;
 - **SODA** services deliver spatial and spectral cutouts;
